## Day 7 - Outlier Detection and Data Distribution

```python
import pandas as pd

url = "https://raw.githubusercontent.com/shreeshreeya69-crypto/simple-linear-regression-ybi/main/Advertising.csv"

df = pd.read_csv(url)

df = df.drop(columns=['Unnamed: 0'], errors='ignore')

df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df["TV"], bins=20)

plt.xlabel("TV Advertising")
plt.ylabel("Frequency")
plt.title("Distribution of TV Advertising")

plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["sales"], bins=20)

plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.title("Distribution of Sales")

plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.boxplot(df["TV"])

plt.ylabel("TV Advertising")
plt.title("Boxplot of TV Advertising")

plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.boxplot(df["sales"])

plt.ylabel("Sales")
plt.title("Boxplot of Sales")

plt.show()
```

```python
Q1 = df["TV"].quantile(0.25)
Q3 = df["TV"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
tv_outliers = df[(df["TV"] < lower_limit) | (df["TV"] > upper_limit)]

print("Number of possible TV outliers:", len(tv_outliers))
```

```python
Q1_sales = df["sales"].quantile(0.25)
Q3_sales = df["sales"].quantile(0.75)

IQR_sales = Q3_sales - Q1_sales

lower_sales = Q1_sales - 1.5 * IQR_sales
upper_sales = Q3_sales + 1.5 * IQR_sales

print("Q1:", Q1_sales)
print("Q3:", Q3_sales)
print("IQR:", IQR_sales)
print("Lower Limit:", lower_sales)
print("Upper Limit:", upper_sales)
```

```python
sales_outliers = df[
    (df["sales"] < lower_sales) |
    (df["sales"] > upper_sales)
]

print("Number of possible Sales outliers:", len(sales_outliers))
```

## Day 7 - Observation

During Day 7, the distribution of the selected variables was examined using histograms and boxplots. The distribution of TV advertising expenditure and Sales was visualized to understand the spread and concentration of the observations.

Boxplots were used to identify potential outliers in the TV advertising and Sales variables. The Interquartile Range (IQR) method was also applied to determine the lower and upper limits for identifying possible outliers.

The analysis helps in understanding the overall structure and variability of the data before developing the Simple Linear Regression model.

Potential outliers were identified for further consideration; however, they were not removed at this stage because an unusual observation does not necessarily represent an incorrect value. Further analysis will determine whether any treatment is required.

The dataset is therefore considered suitable for continuing with the next stages of the regression project.
